## Skin Issue

Mapping the interior of the cloud is something Squarnet handles with a high accuracy.
However, preserving the surface structure is where the real challenge begins.

## Carthesian Heuristics 🧊

Squarnet fitting method is gounded on a sorting method, applyed on Carthesian heuristics: \
We want the x, y, z.. coordinate to be increasing along the row, column... axis of the grid 📈. \
It will work perfectly well for any cloud which is:
- not too peaky 💥 (besides if the peaks are structured like the peaks of the carthesian square, but it would be a coincidence)
- not too thiny 🥥. Then the right model would be a manyfold.
- not too holy 🧀. Squarenet can manage some holls (e.g. in a ring), but not too many of them
- not too heterogeneous. The heuristic will deal with variable intensity. But at some point, almost empty regions are like holes


While a quick fix is to first complete your cloud to a homogenous square, then fit, then replace synthetic points with nans, then you will loose bijectivity. And you might turn into other skin troubles, as points of the surface my get lonely in a very hostile region full of NaN's
